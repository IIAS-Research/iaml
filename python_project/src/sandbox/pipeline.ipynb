```python
import sys
sys.path.insert(0, '../')

import pandas as pd

from automed import *
```

```python
titanic = pd.read_csv('../perf_logger/tests_data/titanic.csv', delimiter=';')

dataset = automed.Dataset(titanic)
dataset.set_label('label')

autom = automed.AutoMed(dataset=dataset)
```

```python
print(autom.debug_load())
print(autom.json_pipeline())
```

```python
pipeline = {
    'step': 'MetaOrderedStep',
    'children': [{
        'step': 'RandomSplit',
        'configuration': {
            'ratio': {
                'value': 0.3
            }
        }},
        {
            'step': 'MetaStep',
            'tag': 'cleaning'
        },{
            'step': 'MetaExplorerStep',
            'tag': 'learning'
        }]
}

autom.load_pipeline(pipeline)
print(autom.json_pipeline())
```

```python
results = autom.run()
[ (r.model, r.compute()) for r in results ]
```

```python
pipeline_test = autom.json_pipeline()
print(pipeline_test)

autom_test = automed.AutoMed(dataset=dataset)
autom_test.load_pipeline(pipeline_test)
print(autom_test.json_pipeline())

results_test = autom_test.run()
[ (r.model, r.compute()) for r in results_test ]
```
