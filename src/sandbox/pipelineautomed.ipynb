```python
import sys
sys.path.insert(0, '../')

import pandas as pd

from automed import *
```

```python
titanic = pd.read_csv('../perf_logger/tests_data/titanic.csv', delimiter=';')
#nlp = pd.read_csv('../perf_logger/tests_data/nlp_research_paper.csv', delimiter=',')

dataset = automed.Dataset(titanic.iloc[:200].copy())
#nlp = automed.Dataset(nlp.iloc[:200].copy())

dataset.set_label('label')
#nlp.disable_columns(['label_2', 'label_3', 'label_4', 'label_5'])
#nlp.set_label(['label_0', 'label_1','label_2', 'label_3', 'label_4', 'label_5'])
#print(dataset.is_multilabel)
#print(nlp.is_multilabel)
#print(nlp.X_train.shape)
#print(nlp.X_train.columns)
#print(nlp.y_train.shape)
#print(nlp.y_train.columns)
autom = automed.AutoMed(dataset=dataset)
```

```python
print(autom.debug_load())
print(autom.json_pipeline())
```

```python
pipeline = {
    'step': 'MetaOrderedStep',
    'children': [{
        'step': 'RandomSplit',
        'configuration': {
            'ratio': {
                'value': 0.3
            }
        }},
        {
            'step': 'MetaStep',
            'tag': 'cleaning'
        },
        {
            'step': 'MetaStep',
            'tag': 'metric'
        },
        {
            'step': 'MetaExplorerStep',
            'tag': 'learning'
        }]
}

autom.load_pipeline(pipeline)
```
