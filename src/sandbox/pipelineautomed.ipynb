```python
from sklearn.utils.multiclass import type_of_target

titanic = pd.read_csv('../perf_logger/tests_data/IMDB-Dataset.csv', delimiter=',')
print(titanic.shape)
#nlp = pd.read_csv('../perf_logger/tests_data/nlp_research_paper.csv', delimiter=',')
#drug = pd.read_csv('../perf_logger/tests_data/Drug_Classification.csv', delimiter = ',')
#print(type_of_target(titanic['category']))
dataset = automed.Dataset(titanic.iloc[:800].copy())
#nlp = automed.Dataset(nlp.iloc[:100].copy())
#drug = automed.Dataset(drug.iloc[:200].copy())

dataset.set_label('sentiment')
#dataset.set_label(['label_0', 'label_1','label_2', 'label_3', 'label_4', 'label_5'])
#dataset.disable_column('TITLE')
#drug.set_label('label')
#print(drug.X_train)
#print(drug.y_train)

print(dataset.is_multilabel)
#print(nlp.is_multilabel)
#print(drug.is_multilabel)

#print(dataset.X_train.shape)
#print(dataset.X_train.columns)
#print(dataset.y_train.shape)
#print(dataset.y_train.columns)
#print(type_of_target(dataset.y_train))


autom = automed.AutoMed(dataset=dataset)
```

```python
print(autom.debug_load())
print(autom.json_pipeline())
```

```python
pipeline = {
    'step': 'MetaOrderedStep',
    'children': [{
        'step': 'RandomSplit',
        'configuration': {
            'ratio': {
                'value': 0.3
            }
        }},
        {
            'step': 'MetaStep',
            'tag': 'cleaning'
        },
        {
            'step': 'MetaStep',
            'tag': 'metric'
        },
        {
            'step': 'MetaExplorerStep',
            'tag': 'learning'
        }]
}

autom.load_pipeline(pipeline)
```
