```python
# Importation des bibliothèques nécessaires
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.preprocessing import MultiLabelBinarizer
from sklearn.multioutput import MultiOutputClassifier
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score
import re
import nltk
from nltk.stem import PorterStemmer
from nltk.tokenize import word_tokenize
```

```python
nltk.download('punkt_tab')
```

```python
stemmer = PorterStemmer()
```

```python
#nlp = pd.read_csv('../perf_logger/tests_data/nlp_research_paper.csv', delimiter=',')
nlp = pd.read_csv('../perf_logger/tests_data/bbc-text.csv', delimiter=',')
```

```python
nlp.shape
```

```python
def preprocess(text):
    text = text.lower() # Conversion en minuscules
    text = re.sub('[^\w\s]', '', text) # Suppression de la ponctuation
    tokens = word_tokenize(text) #Tokenization
    stemmed_tokens = [stemmer.stem(token) for token in tokens] #Stemming permet de réduire les mots à leur racine, cela peut aider à améliorer les performances de la classification en réduisant la complexité
    return ' '.join(stemmed_tokens) # Retourne le texte nettoyé et stemmé
```

```python
nlp['processed_text'] = nlp['text'].apply(preprocess)
```

```python
labels = nlp['text']
mlb = MultiLabelBinarizer()
labels_binarized = mlb.fit_transform(labels.values)
```

```python
labels_binarized
```

```python
X_train, X_test, y_train, y_test = train_test_split(nlp['text'], labels_binarized, test_size = 0.2, random_state= 42)
print('X_train shape:', X_train.shape)
print('X_test shape :', X_test.shape)
print('y_train shape:', y_train.shape)
print('y_test shape:', y_test.shape)
```

```python
# CounVecrorizer est une technique de prétraitement de texte courament utilisée dans les tâches de traitement du langage naturel NLP pour convertir une collection de documents textuels en une representation numerique
# CounVectorize procède à la tokenization des données textuelles et au comptage des occurences de chaque token, il crée ensuite une matrice dont les lgnes reprtesentent les documents et les colonnes les mots-clés. Les valeurs cellules indiquent la fréquence de chaque mo clé dans chaque document. Cette matrice est connue sous le nom de matrice document terme


vectorizer = CountVectorizer(stop_words = 'english')
X_train_counts = vectorizer.fit_transform(X_train)
X_test_counts = vectorizer.transform(X_test)
```

```python
X_train_counts.shape
```

```python
clf = MultiOutputClassifier(MultinomialNB())
clf.fit(X_train_counts, y_train)
y_pred = clf.predict(X_test_counts)
```

```python
y_pred
```

```python
from sklearn.metrics import *
print('accuracy score:', accuracy_score(y_test, y_pred))
print('precision score:', precision_score(y_test, y_pred, average = 'samples'))
print('recall score:', recall_score(y_test, y_pred, average= 'samples'))
print('f1 score :', f1_score(y_test, y_pred, average = 'samples'))
```
