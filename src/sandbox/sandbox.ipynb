```python
from sandbox import TimedPoolExecutor
import time
```

```python
def test(timeout, text):
    time.sleep(timeout)
    return "OK", text
```

```python
results = []
with TimedPoolExecutor(max_workers=4) as executor:

    executor.submit(test, 10, 1)
    executor.submit(test, 3, 2)
    executor.submit(test, 3, 3)
    executor.submit(test, 3, 4)

    executor.submit(test, 3, 5)
    executor.submit(test, 3, 6)
    executor.submit(test, 3, 7)
    executor.submit(test, 3, 8)

    executor.submit(test, 3, 9)
    executor.submit(test, 3, 10)
    executor.submit(test, 3, 11)
    executor.submit(test, 3, 12)

    executor.submit(test, 10, 13)
    executor.submit(test, 20, 5)

    results = executor.join(8)
    
results
```
